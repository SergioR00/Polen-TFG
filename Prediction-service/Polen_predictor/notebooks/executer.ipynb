```python
import papermill as pm
import platform
from pathlib import Path

# 1. Detectamos de forma segura dónde está este script/notebook 'executer'
try:
    base_dir = Path(__file__).resolve().parent
except NameError:
    base_dir = Path('.').resolve()

# 2. Forzamos la creación de la carpeta 'my_datasets' si no existe
# Esto evita el error de Pandas en Docker de "non-existent directory"
datasets_dir = base_dir.parent / "my_datasets"
datasets_dir.mkdir(parents=True, exist_ok=True)

notebooks = [
    "get_Old_Data.ipynb",
    "update_Data_&_predictions.ipynb",
    "add_Features.ipynb",
    "predict_&_upload.ipynb"
]

# Definimos el dispositivo nulo según el Sistema Operativo
null_device = "NUL" if platform.system() == "Windows" else "/dev/null"

for nb in notebooks:
    # Construimos la ruta absoluta al notebook actual
    ruta_absoluta_nb = base_dir / nb
    
    print(f"Ejecutando {nb} de forma universal...")
    
    try:
        pm.execute_notebook(
            str(ruta_absoluta_nb),
            null_device,
            parameters=dict(proceso="silencioso"),
            # LECCIÓN CLAVE: Le decimos a papermill que ejecute el notebook 
            # parándose firmemente en la carpeta de los notebooks.
            cwd=str(base_dir) 
        )
    except Exception as e:
        print(f"Error al ejecutar {nb}: {e}")
        # Opcional: puedes meter un raise aquí si quieres que todo se detenga si uno falla
```
