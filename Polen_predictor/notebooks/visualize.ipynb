```python
import pandas as pd
import matplotlib.pyplot as plt

tipos_polinicos = ["gramineas", "cupresaceas", "olivo", "platano", "urticaceas", "quenopodiaceas"]

for tipo in tipos_polinicos:
    # 1. Cargar datos
    df = pd.read_csv(f'../new_datasets/datos_{tipo}.csv')
    
    # 2. Convertir fecha y filtrar desde 2014
    df['fecha'] = pd.to_datetime(df['fecha'])
    df = df[df['fecha'].dt.year >= 2014]
    
    # 3. Graficar
    plt.figure(figsize=(12, 4))
    plt.plot(df['fecha'], df['granos_de_polen_x_metro_cubico'], label=tipo)
    
    plt.title(f"Polen: {tipo}")
    plt.ylabel("Granos / $m^3$")
    plt.xlabel("Año")
    plt.legend()
    plt.show()
```
